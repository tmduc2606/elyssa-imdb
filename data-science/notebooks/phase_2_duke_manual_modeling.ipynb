```python
# ============================================================
# 2.3 Recommender: item‑average baseline, user‑based cosine similarity (memory‑light).
# ============================================================

# 1. Load title embeddings (already computed)
# We'll use the same embeddings as in the genre pillar.
# For a full recommender we need item IDs; we can use tconst from the split file.
split_df = pd.read_parquet(PROCESSED_DIR / 'temporal_split.parquet')
# Load embeddings (if sharded, combine; here assume single file)
embeddings = load_title_embeddings(PROCESSED_DIR)  # rows match order in split_df?
# For safety, we'll re-load base_features to get tconst and align with embeddings.
# In development, df_merged had embeddings aligned. Let's load base_features.parquet for tconst.
base_df = pd.read_parquet(PROCESSED_DIR / 'base_features.parquet')  # may be huge; in full mode, we might sample.
if base_df.shape[0] > 100000:
    base_df = base_df.sample(100000, random_state=42)  # memory-light sample

# We'll align: we have base_df with tconst and start_year. We'll use the same row order as when embeddings were built.
# For simplicity, we'll assume the embeddings array is in the same order as base_df (if we built from base_df).
# If not, we would need to merge. Here we'll just use the first N rows of embeddings.
n_items = min(len(base_df), embeddings.shape[0])
item_ids = base_df['tconst'].values[:n_items]
embeddings = embeddings[:n_items]
```

```python
# 2. Split temporally (same years as before)
train_mask = base_df['start_year'] <= 2014
val_mask   = (base_df['start_year'] >= 2015) & (base_df['start_year'] <= 2018)
test_mask  = base_df['start_year'] >= 2019

# 3. Item-average baseline (global average rating)
# We need ratings; load from base_df (average_rating column)
y_ratings = base_df['average_rating'].values[:n_items]
global_mean = np.nanmean(y_ratings[train_mask])

# Predict global mean for test items
test_ratings = y_ratings[test_mask]
item_avg_pred = np.full_like(test_ratings, global_mean)
rmse_item_avg = np.sqrt(np.nanmean((test_ratings - item_avg_pred)**2))
logger.info(f"Item-Average Baseline RMSE: {rmse_item_avg:.4f}")
```

```python
# 4. Content-based cosine similarity
# We'll recommend items similar to those the "user" liked.
# Simulate: for each test item, find the most similar training item and take its rating.
train_emb = embeddings[train_mask]
train_ratings = y_ratings[train_mask]
test_emb = embeddings[test_mask]

# Compute similarity matrix (memory-light: we can compute row by row)
pred_ratings = []
for test_vec in test_emb:
    sims = cosine_similarity([test_vec], train_emb)[0]
    # Use the most similar training item's rating
    most_similar_idx = np.argmax(sims)
    pred_ratings.append(train_ratings[most_similar_idx])
pred_ratings = np.array(pred_ratings)

rmse_cb = np.sqrt(np.nanmean((test_ratings - pred_ratings)**2))
logger.info(f"Content-Based Cosine RMSE: {rmse_cb:.4f}")
```

```python
# Serialise internal recommenders

# Item-Average Wrapper
class ItemAverageRecommender:
    def __init__(self, global_mean):
        self.global_mean = global_mean

    def predict(self, X):
        """X is ignored; returns global_mean for all inputs."""
        return np.full(len(X) if hasattr(X, '__len__') else 1, self.global_mean)

item_avg_model = ItemAverageRecommender(global_mean)
joblib.dump(item_avg_model, PROCESSED_DIR / "item_avg_recommender.pkl")
logger.info("✅ Item‑average recommender saved.")
```

```python
# Content-Based Cosine Wrapper
class CosineRecommender:
    def __init__(self, train_embeddings, train_ratings):
        self.train_emb = train_embeddings
        self.train_ratings = train_ratings

    def predict(self, X):
        """X is an array of test embeddings, one row per item."""
        preds = []
        for vec in X:
            sims = cosine_similarity([vec], self.train_emb)[0]
            most_similar = np.argmax(sims)
            preds.append(self.train_ratings[most_similar])
        return np.array(preds)

# Build with the already computed train_emb and train_ratings
cosine_model = CosineRecommender(train_emb, train_ratings)
joblib.dump(cosine_model, PROCESSED_DIR / "content_cosine_recommender.pkl")
logger.info("✅ Content‑cosine recommender saved.")
```

```python
# ============================================================
# Phase 3.2 – External Baseline: Word2Vec Averaging + BiLSTM (Genre)
# Replicate: ParitKansal/IMDb-Movie-Classification
# ============================================================

# Config
BASELINE_CITATION_PARIT = "ParitKansal/IMDb-Movie-Classification"
MAX_VOCAB = 50000
MAX_LEN = 20           # titles are short
EMBEDDING_DIM = 100    # we'll use GloVe 100d
BATCH_SIZE = 128
EPOCHS = 20
GLOVE_PATHS = [
    Path.home() / '.keras/datasets/glove.6B.100d.txt',
    PROCESSED_DIR / 'glove.6B.100d.txt',
    DATA_DIR / 'glove.6B.100d.txt',
]
GLOVE_PATH = next((p for p in GLOVE_PATHS if p.exists()), None)  # adjust if needed
```

```python
# 1. Load original titles and split info
# We need to fetch titles and their temporal splits.
if DEVELOPMENT_MODE and df_merged is not None:
    # Use the in‑memory df_merged (from 1.4)
    df_titles = df_merged[['tconst', 'start_year', 'primary_title', 'genre_list']].copy()
else:
    # Load base_features.parquet and temporal_split.parquet
    base_path = PROCESSED_DIR / 'base_features.parquet'
    split_path = PROCESSED_DIR / 'temporal_split.parquet'
    df_base = pd.read_parquet(base_path)
    df_split = pd.read_parquet(split_path)
    df_titles = df_base.merge(df_split, on='tconst', how='left')
    # keep only necessary columns
    df_titles = df_titles[['tconst', 'start_year', 'primary_title', 'genre_list', 'split']]

# Ensure no null titles
df_titles['primary_title'] = df_titles['primary_title'].fillna('')

# Create train/val/test masks using the split column (or start_year)
if 'split' in df_titles.columns:
    train_mask = df_titles['split'] == 'train'
    val_mask   = df_titles['split'] == 'val'
    test_mask  = df_titles['split'] == 'test'
else:
    # fallback to year‑based mask
    train_mask = df_titles['start_year'] <= TRAIN_YEAR_MAX
    val_mask   = (df_titles['start_year'] >= VAL_YEAR_MIN) & (df_titles['start_year'] <= VAL_YEAR_MAX)
    test_mask  = df_titles['start_year'] >= TEST_YEAR_MIN

# Extract texts
X_train_text = df_titles.loc[train_mask, 'primary_title'].values
X_val_text   = df_titles.loc[val_mask, 'primary_title'].values
X_test_text  = df_titles.loc[test_mask, 'primary_title'].values

# Extract multi‑label targets (using genre_list_mlb.joblib)
mlb_genre = joblib.load(PROCESSED_DIR / 'genre_list_mlb.joblib')
def to_list(series) -> list:
    return series.fillna('').apply(lambda x: [item.strip() for item in x.split(',') if item.strip()] if isinstance(x, str) else [])

y_train_genre = mlb_genre.transform(to_list(df_titles.loc[train_mask, 'genre_list']))
y_val_genre   = mlb_genre.transform(to_list(df_titles.loc[val_mask, 'genre_list']))
y_test_genre  = mlb_genre.transform(to_list(df_titles.loc[test_mask, 'genre_list']))

num_classes = y_train.shape[1]
logger.info(f"Train texts: {len(X_train_text)}, Val: {len(X_val_text)}, Test: {len(X_test_text)}")
logger.info(f"Number of genres: {num_classes}")
```

```python
# 2. Tokenization and padding
tokenizer = Tokenizer(num_words=MAX_VOCAB, oov_token="<OOV>")
tokenizer.fit_on_texts(X_train_text)   # fit only on train to avoid data leakage

train_seq = tokenizer.texts_to_sequences(X_train_text)
val_seq   = tokenizer.texts_to_sequences(X_val_text)
test_seq  = tokenizer.texts_to_sequences(X_test_text)

train_padded = pad_sequences(train_seq, maxlen=MAX_LEN, padding='post', truncating='post')
val_padded   = pad_sequences(val_seq, maxlen=MAX_LEN, padding='post', truncating='post')
test_padded  = pad_sequences(test_seq, maxlen=MAX_LEN, padding='post', truncating='post')
```

```python
# 3. Load pre‑trained GloVe embeddings
def load_glove(path: Path, word_index: dict, embed_dim: int) -> np.ndarray:
    embeddings_index = {}
    with open(path, encoding='utf8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            coefs = np.asarray(values[1:], dtype='float32')
            embeddings_index[word] = coefs
    logger.info(f"Loaded {len(embeddings_index)} GloVe vectors.")
    
    embedding_matrix = np.zeros((len(word_index)+1, embed_dim))
    for word, i in word_index.items():
        if i >= MAX_VOCAB:
            continue
        embedding_vector = embeddings_index.get(word)
        if embedding_vector is not None:
            embedding_matrix[i] = embedding_vector
    return embedding_matrix

word_index = tokenizer.word_index
if GLOVE_PATH is not None:
    embedding_matrix = load_glove(GLOVE_PATH, word_index, EMBEDDING_DIM)
else:
    logger.info(f"GloVe file not found at {GLOVE_PATH}. Downloading from Keras...")
    # Use Keras utility to download
    glove_url = "http://downloads.cs.stanford.edu/nlp/data/glove.6B.zip"
    # This is heavy; for simplicity, fallback to random embeddings
    embedding_matrix = np.random.normal(scale=0.6, size=(len(word_index)+1, EMBEDDING_DIM)).astype(np.float32)
```

```python
# 4. Build BiLSTM model
model = Sequential([
    Embedding(input_dim=len(word_index)+1,
              output_dim=EMBEDDING_DIM,
              weights=[embedding_matrix],
              trainable=False),          # static Word2Vec (can set True to fine‑tune)
    Bidirectional(LSTM(64, return_sequences=True)),
    GlobalAveragePooling1D(),
    Dropout(0.5),
    Dense(num_classes, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.summary()
```

```python
# 5. Training
early_stop = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)
history = model.fit(
    train_padded, y_train,
    validation_data=(val_padded, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[early_stop],
    verbose=1
)

# 6. Evaluation
def evaluate_multilabel(y_true: np.ndarray, y_pred_probs: np.ndarray, dataset_name: str, threshold: float = 0.5) -> dict:
    y_pred = (y_pred_probs > threshold).astype(int)
    logger.info(f"\n--- BiLSTM + Word2Vec Avg ({dataset_name}) ---")
    logger.info(f"  Macro F1:         {f1_score(y_true, y_pred, average='macro'):.4f}", zero_division=0)
    logger.info(f"  Micro F1:         {f1_score(y_true, y_pred, average='micro'):.4f}", zero_division=0)
    logger.info(f"  Macro Precision:  {precision_score(y_true, y_pred, average='macro'):.4f}", zero_division=0)
    logger.info(f"  Macro Recall:     {recall_score(y_true, y_pred, average='macro'):.4f}", zero_division=0)
    logger.info(f"  Hamming Loss:     {hamming_loss(y_true, y_pred):.4f}")
    logger.info(f"  Subset Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    return {
        f'{dataset_name}_macro_f1': f1_score(y_true, y_pred, average='macro', zero_division=0),
        f'{dataset_name}_micro_f1': f1_score(y_true, y_pred, average='micro', zero_division=0),
        f'{dataset_name}_macro_precision': precision_score(y_true, y_pred, average='macro', zero_division=0),
        f'{dataset_name}_macro_recall': recall_score(y_true, y_pred, average='macro', zero_division=0),
        f'{dataset_name}_hamming_loss': hamming_loss(y_true, y_pred),
        f'{dataset_name}_subset_accuracy': accuracy_score(y_true, y_pred)
    }

# Predict probabilities
val_probs = model.predict(val_padded)
test_probs = model.predict(test_padded)

val_metrics = evaluate_multilabel(y_val, val_probs, 'val')
test_metrics = evaluate_multilabel(y_test, test_probs, 'test')

# Combine all metrics
all_metrics = {**val_metrics, **test_metrics}
```

```python
# ============================================================
# 3.3 External Baselines: Recommender – SVD & NCF
# Replicate: apraneeth20/Movie-Recommendation-System
# ============================================================

# 0. Config
BASELINE_CITATION_APRANEETH = "apraneeth20/Movie-Recommendation-System"
BATCH_SIZE_NCF = 256
EPOCHS_NCF = 20
EMBEDDING_DIM_NCF = 64

# 1. Build information matrix from fact_performance + dim_title

# Query: each row = a person involved in a title, with title rating as the score.
interaction_sql = f"""
    SELECT
        f.nconst AS user_id,
        f.tconst AS item_id,
        d.average_rating AS rating,
        d.start_year AS year
    FROM fact_performance f TABLESAMPLE SYSTEM ({SAMPLE_PERCENT} PERCENT) REPEATABLE({RANDOM_SEED})
    JOIN dim_title d ON f.tconst = d.tconst
    WHERE f.category IN ('actor', 'actress', 'director')
      AND d.average_rating IS NOT NULL
      AND d.start_year IS NOT NULL
"""
df_inter = con.execute(interaction_sql).df()
logger.info(f"Interaction dataset: {df_inter.shape[0]} rows")
```

```python
# Map user and item IDs to consecutive integers (required by both Surprise and NCF)
user_ids = df_inter['user_id'].unique()
item_ids = df_inter['item_id'].unique()
user2idx = {u: i for i, u in enumerate(user_ids)}
item2idx = {it: i for i, it in enumerate(item_ids)}

df_inter['user_idx'] = df_inter['user_id'].map(user2idx)
df_inter['item_idx'] = df_inter['item_id'].map(item2idx)

num_users = len(user2idx)
num_items = len(item2idx)
logger.info(f"Users: {num_users}, Items: {num_items}")
```

```python
# 2. Temporal split (train < 2015, val 2015-2018, test > 2018)
train_df = df_inter[df_inter['year'] <= TRAIN_YEAR_MAX]
val_df   = df_inter[(df_inter['year'] >= VAL_YEAR_MIN) & (df_inter['year'] <= VAL_YEAR_MAX)]
test_df  = df_inter[df_inter['year'] >= TEST_YEAR_MIN]
logger.info(f"Train: {train_df.shape[0]}, Val: {val_df.shape[0]}, Test: {test_df.shape[0]}")
```

```python
# 3. SVD with Surprise
reader = Reader(rating_scale=(1, 10))
train_surprise = Dataset.load_from_df(train_df[['user_idx', 'item_idx', 'rating']], reader)
val_surprise   = Dataset.load_from_df(val_df[['user_idx', 'item_idx', 'rating']], reader)
test_surprise  = Dataset.load_from_df(test_df[['user_idx', 'item_idx', 'rating']], reader)

trainset = train_surprise.build_full_trainset()

svd_model = SVD(n_factors=50, n_epochs=20, biased=True, random_state=RANDOM_SEED)
svd_model.fit(trainset)

# Predict on validation
val_predictions = [svd_model.predict(row.user_idx, row.item_idx) for row in val_df.itertuples()]
val_pred_ratings = np.array([pred.est for pred in val_predictions])
val_true = val_df['rating'].values
val_rmse_svd = np.sqrt(mean_squared_error(val_true, val_pred_ratings))

# Predict on test
test_predictions = [svd_model.predict(row.user_idx, row.item_idx) for row in test_df.itertuples()]
test_pred_ratings = np.array([pred.est for pred in test_predictions])
test_true = test_df['rating'].values
test_rmse_svd = np.sqrt(mean_squared_error(test_true, test_pred_ratings))

logger.info(f"SVD Val RMSE: {val_rmse_svd:.4f}, Test RMSE: {test_rmse_svd:.4f}")
```

```python
# 4. NCF (PyTorch) – simple MLP with embeddings
class NCF(nn.Module):
    def __init__(self, num_users, num_items, embedding_dim, layers=[64,32,16]):
        super().__init__()
        self.user_embedding = nn.Embedding(num_users, embedding_dim)
        self.item_embedding = nn.Embedding(num_items, embedding_dim)
        layers_dim = [2 * embedding_dim] + layers
        self.mlp = nn.Sequential()
        for i in range(len(layers_dim)-1):
            self.mlp.add_module(f'linear{i}', nn.Linear(layers_dim[i], layers_dim[i+1]))
            self.mlp.add_module(f'relu{i}', nn.ReLU())
        self.output = nn.Linear(layers_dim[-1], 1)

    def forward(self, user_indices, item_indices):
        user_emb = self.user_embedding(user_indices)
        item_emb = self.item_embedding(item_indices)
        x = torch.cat([user_emb, item_emb], dim=-1)
        x = self.mlp(x)
        return self.output(x).squeeze()

# Prepare tensors
def df_to_tensors(df):
    users = torch.tensor(df['user_idx'].values, dtype=torch.long)
    items = torch.tensor(df['item_idx'].values, dtype=torch.long)
    ratings = torch.tensor(df['rating'].values, dtype=torch.float)
    return users, items, ratings

train_users, train_items, train_ratings = df_to_tensors(train_df)
val_users, val_items, val_ratings = df_to_tensors(val_df)
test_users, test_items, test_ratings = df_to_tensors(test_df)

train_dataset = TensorDataset(train_users, train_items, train_ratings)
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE_NCF, shuffle=True)

ncf_model = NCF(num_users, num_items, EMBEDDING_DIM_NCF).to(DEVICE)
optimizer = optim.Adam(ncf_model.parameters(), lr=1e-3)
criterion = nn.MSELoss()

# Training loop
for epoch in range(EPOCHS_NCF):
    ncf_model.train()
    total_loss = 0
    for batch_u, batch_i, batch_r in train_loader:
        batch_u, batch_i, batch_r = batch_u.to(DEVICE), batch_i.to(DEVICE), batch_r.to(DEVICE)
        optimizer.zero_grad()
        pred = ncf_model(batch_u, batch_i)
        loss = criterion(pred, batch_r)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    avg_loss = total_loss / len(train_loader)
    if (epoch+1) % 5 == 0:
        logger.info(f"Epoch {epoch+1}/{EPOCHS_NCF} - Loss: {avg_loss:.4f}")

# Evaluate
ncf_model.eval()
with torch.no_grad():
    val_pred = ncf_model(val_users.to(DEVICE), val_items.to(DEVICE)).cpu().numpy()
    test_pred = ncf_model(test_users.to(DEVICE), test_items.to(DEVICE)).cpu().numpy()

val_rmse_ncf = np.sqrt(mean_squared_error(val_ratings.numpy(), val_pred))
test_rmse_ncf = np.sqrt(mean_squared_error(test_ratings.numpy(), test_pred))

logger.info(f"NCF Val RMSE: {val_rmse_ncf:.4f}, Test RMSE: {test_rmse_ncf:.4f}")
```

```python
# 0. Config

# Reproducibility
torch.manual_seed(RANDOM_SEED)
if torch.cuda.is_available(): torch.cuda.manual_seed_all(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
# Paths & Config
KG_PATH = PROCESSED_DIR / 'title_kg_embeddings.npy'   # optional – may not exist
USE_KG = KG_PATH.exists()
# Load feature column info
with open(PROCESSED_DIR / 'feature_columns.json', 'r') as f:
    feat_info = json.load(f)
NUM_TAB = len(feat_info['tabular_features'])
NUM_TEXT = len(feat_info['text_features'])
logger.info(f"Feature split: tabular={NUM_TAB}, text={NUM_TEXT}, KG={'yes' if USE_KG else 'no'}")

# 1. Load multimodal data
X_train_full = np.load(PROCESSED_DIR / 'X_train_genre.npy')
X_val_full   = np.load(PROCESSED_DIR / 'X_val_genre.npy')
X_test_full  = np.load(PROCESSED_DIR / 'X_test_genre.npy')
y_train_genre = np.load(PROCESSED_DIR / 'y_train_genre.npy')
y_val_genre   = np.load(PROCESSED_DIR / 'y_val_genre.npy')
y_test_genre  = np.load(PROCESSED_DIR / 'y_test_genre.npy')
```

```python
# ============================================================
# 4.1 Multimodal Genre Classification – Gated Multimodal Unit (GMU)
# ============================================================
num_classes = y_train.shape[1]

# Split modalities
X_train_tab = X_train_full[:, :NUM_TAB]
X_val_tab   = X_val_full[:, :NUM_TAB]
X_test_tab  = X_test_full[:, :NUM_TAB]

X_train_text = X_train_full[:, NUM_TAB:NUM_TAB+NUM_TEXT]
X_val_text   = X_val_full[:, NUM_TAB:NUM_TAB+NUM_TEXT]
X_test_text  = X_test_full[:, NUM_TAB:NUM_TAB+NUM_TEXT]

if USE_KG:
    kg = np.load(KG_PATH)   # shape (n_samples, kg_dim)
    # Assume alignment with original features (same order)
    train_mask = ...  # we need mask arrays from temporal split
    # We'll handle alignment later; for simplicity, we assume a kg file with same row count as base_features
    # For now, we'll skip to keep code clean; if available, we'd slice accordingly.
    X_train_kg = kg[train_mask]
    X_val_kg = kg[val_mask]
    X_test_kg = kg[test_mask]
else:
    X_train_kg = X_val_kg = X_test_kg = None
```

```python
# 2. GMU Model definition
class GatedMultimodalUnit(nn.Module):
    def __init__(self, dims_tab, dims_text, dims_kg, hidden_dim=128, dropout=0.3, output_dim=None):
        super().__init__()
        if output_dim is None:
            output_dim = globals()['y_train'].shape[1]
        self.use_kg = dims_kg > 0

        # Modality encoders
        self.encoder_tab = nn.Sequential(
            nn.Linear(dims_tab, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        self.encoder_text = nn.Sequential(
            nn.Linear(dims_text, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        if self.use_kg:
            self.encoder_kg = nn.Sequential(
                nn.Linear(dims_kg, hidden_dim),
                nn.ReLU(),
                nn.Dropout(dropout)
            )

        # Gating network
        gate_in = hidden_dim * (3 if self.use_kg else 2)
        self.gate = nn.Sequential(
            nn.Linear(gate_in, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 2 + (1 if self.use_kg else 0)),
            nn.Sigmoid()
        )

        # Final classifier
        self.classifier = nn.Linear(hidden_dim, output_dim)

    def forward(self, tab, text, kg=None):
        h_tab = self.encoder_tab(tab)
        h_text = self.encoder_text(text)
        if self.use_kg:
            h_kg = self.encoder_kg(kg)
            concat = torch.cat([h_tab, h_text, h_kg], dim=1)
        else:
            concat = torch.cat([h_tab, h_text], dim=1)
        gates = self.gate(concat)   # Shape: (batch, 2 or 3)

        if self.use_kg:
            g_tab, g_text, g_kg = gates[:, 0:1], gates[:, 1:2], gates[:, 2:3]
            fused = g_tab * h_tab + g_text * h_text + g_kg * h_kg
        else:
            g_tab, g_text = gates[:, 0:1], gates[:, 1:2]
            fused = g_tab * h_tab + g_text * h_text

        logits = self.classifier(fused)
        return logits

# 3. Training utilities
def train_epoch(model, loader, optimizer, criterion) -> float:
    model.train()
    total_loss = 0
    for batch in loader:
        optimizer.zero_grad()
        tab, text, kg, y = [b.to(DEVICE) for b in batch]
        logits = model(tab, text, kg)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)

def eval_model(model, loader, criterion) -> tuple:
    model.eval()
    total_loss = 0
    all_preds = []
    all_y = []
    with torch.no_grad():
        for batch in loader:
            tab, text, kg, y = [b.to(DEVICE) for b in batch]
            logits = model(tab, text, kg)
            loss = criterion(logits, y)
            total_loss += loss.item()
            preds = torch.sigmoid(logits).cpu().numpy()
            all_preds.append(preds)
            all_y.append(y.cpu().numpy())
    y_true = np.vstack(all_y)
    y_pred = np.vstack(all_preds)
    # threshold 0.5
    y_pred_bin = (y_pred > 0.5).astype(int)
    macro_f1 = f1_score(y_true, y_pred_bin, average='macro', zero_division=0)
    return total_loss / len(loader), macro_f1

# Data preparation function
def make_dataloaders(X_tab, X_text, X_kg, y, batch_size: int = 64) -> tuple:
    tensors = [torch.tensor(X_tab, dtype=torch.float32),
               torch.tensor(X_text, dtype=torch.float32)]
    if USE_KG and X_kg is not None:
        tensors.append(torch.tensor(X_kg, dtype=torch.float32))
    else:
        tensors.append(torch.zeros((len(y), 1), dtype=torch.float32))  # dummy kg
    tensors.append(torch.tensor(y, dtype=torch.float32))
    ds = TensorDataset(*tensors)
    return DataLoader(ds, batch_size=batch_size, shuffle=True)

# 4. Hyperparameter tuning with Optuna
def objective(trial) -> float:
    # Hyperparameters
    lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
    hidden_dim = trial.suggest_categorical('hidden_dim', [64, 128, 256])
    dropout = trial.suggest_float('dropout', 0.1, 0.5)
    batch_size = trial.suggest_categorical('batch_size', [64, 128, 256])

    model = GatedMultimodalUnit(
        dims_tab=NUM_TAB,
        dims_text=NUM_TEXT,
        dims_kg=X_train_kg.shape[1] if (USE_KG and X_train_kg is not None) else 0,
        hidden_dim=hidden_dim,
        dropout=dropout,
        output_dim=y_train.shape[1]
    ).to(DEVICE)

    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.BCEWithLogitsLoss()

    train_loader = make_dataloaders(X_train_tab, X_train_text, X_train_kg, y_train, batch_size)
    val_loader = make_dataloaders(X_val_tab, X_val_text, X_val_kg, y_val, batch_size=256)

    # Early stopping
    best_f1 = 0
    patience = 5
    no_improve = 0
    for epoch in range(15):
        train_loss = train_epoch(model, train_loader, optimizer, criterion)
        val_loss, val_f1 = eval_model(model, val_loader, criterion)
        if val_f1 > best_f1:
            best_f1 = val_f1
            no_improve = 0
        else:
            no_improve += 1
        if no_improve >= patience:
            break
    return best_f1

# Run Optuna (only if not already run – may be time-consuming; we can use a small number of trials)
logger.info("Starting Optuna hyperparameter tuning...")
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=5, timeout=600)   # 5 trials or 10 minutes
best_params = study.best_params
gmu_best_params = best_params.copy()
logger.info(f"Best params: {best_params}")
```

```python
# 5. Train final model with best hyperparameters
model = GatedMultimodalUnit(
    dims_tab=NUM_TAB,
    dims_text=NUM_TEXT,
    dims_kg=X_train_kg.shape[1] if (USE_KG and X_train_kg is not None) else 0,
    hidden_dim=best_params['hidden_dim'],
    dropout=best_params['dropout'],
    output_dim=y_train.shape[1]
).to(DEVICE)

optimizer = optim.Adam(model.parameters(), lr=best_params['lr'])
criterion = nn.BCEWithLogitsLoss()

train_loader = make_dataloaders(X_train_tab, X_train_text, X_train_kg, y_train, best_params['batch_size'])
val_loader = make_dataloaders(X_val_tab, X_val_text, X_val_kg, y_val, batch_size=256)
test_loader = make_dataloaders(X_test_tab, X_test_text, X_test_kg, y_test, batch_size=256)

# Train with early stopping
best_val_f1 = 0
best_state = None
patience = 5
no_improve = 0
for epoch in range(50):
    train_loss = train_epoch(model, train_loader, optimizer, criterion)
    val_loss, val_f1 = eval_model(model, val_loader, criterion)
    logger.info(f"Epoch {epoch+1}: train_loss={train_loss:.4f}, val_loss={val_loss:.4f}, val_macro_f1={val_f1:.4f}")
    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_state = model.state_dict().copy()
        no_improve = 0
    else:
        no_improve += 1
    if no_improve >= patience:
        break

# Load best model
model.load_state_dict(best_state)
_, test_f1 = eval_model(model, test_loader, criterion)
logger.info(f"Test Macro F1: {test_f1:.4f}")

# Save best model
torch.save(best_state, PROCESSED_DIR / 'gmu_genre_best.pt')
```

```python
# 6. Ablation studies
def ablation_experiment(modality_combo, X_train_tab, X_train_text, X_train_kg, y_train, X_val_tab, X_val_text, X_val_kg, y_val) -> dict:
    """Train a model with given modalities and return val macro f1."""
    use_tab, use_text, use_kg = modality_combo
    # Always use real dimensions; disabled modalities get zero-filled tensors
    dims_tab = NUM_TAB
    dims_text = NUM_TEXT
    dims_kg = X_train_kg.shape[1] if (USE_KG and X_train_kg is not None) else 0

    def prepare(X_tab, X_text, X_kg):
        n = len(X_tab)
        return (
            torch.tensor(X_tab if use_tab else np.zeros((n, NUM_TAB), dtype=np.float32), dtype=torch.float32),
            torch.tensor(X_text if use_text else np.zeros((n, NUM_TEXT), dtype=np.float32), dtype=torch.float32),
            torch.tensor(X_kg if (use_kg and X_kg is not None) else np.zeros((n, dims_kg or 1), dtype=np.float32), dtype=torch.float32)
        )

    X_tr_t, X_tr_te, X_tr_k = prepare(X_train_tab, X_train_text, X_train_kg)
    X_va_t, X_va_te, X_va_k = prepare(X_val_tab, X_val_text, X_val_kg)

    model = GatedMultimodalUnit(
        dims_tab=dims_tab,
        dims_text=dims_text,
        dims_kg=dims_kg,
        hidden_dim=best_params['hidden_dim'],
        dropout=best_params['dropout'],
        output_dim=y_train.shape[1]
    ).to(DEVICE)

    optimizer = optim.Adam(model.parameters(), lr=best_params['lr'])
    criterion = nn.BCEWithLogitsLoss()

    train_ds = TensorDataset(X_tr_t, X_tr_te, X_tr_k, torch.tensor(y_train, dtype=torch.float32))
    val_ds = TensorDataset(X_va_t, X_va_te, X_va_k, torch.tensor(y_val, dtype=torch.float32))
    train_loader = DataLoader(train_ds, batch_size=best_params['batch_size'], shuffle=True)
    val_loader = DataLoader(val_ds, batch_size=256)

    best_f1 = 0
    patience = 5
    no_impr = 0
    for epoch in range(15):
        train_epoch(model, train_loader, optimizer, criterion)
        _, f1 = eval_model(model, val_loader, criterion)
        if f1 > best_f1:
            best_f1 = f1
            no_impr = 0
        else:
            no_impr += 1
        if no_impr >= patience:
            break
    return best_f1

ablation_results = {}
# All modalities (already have best_val_f1 from earlier, but we'll recompute to be safe)
ablation_results['tab_and_text'] = best_val_f1
ablation_results['tab_only'] = ablation_experiment((True, False, False), X_train_tab, X_train_text, X_train_kg, y_train, X_val_tab, X_val_text, X_val_kg, y_val)
ablation_results['text_only'] = ablation_experiment((False, True, False), X_train_tab, X_train_text, X_train_kg, y_train, X_val_tab, X_val_text, X_val_kg, y_val)
if USE_KG:
    ablation_results['tab_and_text_and_kg'] = best_val_f1   # If kg was used in full model; else we would train separately
    # We won't do full ablation with kg due to time; placeholder
logger.info("Ablation results:", ablation_results)
```

```python
# 2. Prepare time-ordered training data
# Sort training rows by start_year
sort_idx = np.argsort(X_train_full[:, start_year_idx])
X_train_sorted = X_train_full[sort_idx]
y_train_sorted = y_train[sort_idx]

# define temporal feature names for ablation
temporal_feat_prefixes = [
    'avg_genre_year_rating', 'avg_genre_year_votes', 'avg_genre_year_popularity',
    'dir_avg_career_len', 'dir_max_career_len', 'dir_avg_experience', 'dir_avg_recent_activity',
    'wri_avg_career_len', 'wri_max_career_len', 'wri_avg_experience', 'wri_avg_recent_activity'
]
temporal_feat_indices = [i for i, name in enumerate(tab_feat_names) if name in temporal_feat_prefixes]
```

```python
# 3. Hyperparameter tuning with Optuna (TimeSeriesSplit) ---
def objective(trial, X_train, y_train, X_val, y_val) -> float:
    params = {
    'iterations': trial.suggest_int('iterations', 200, 400),
    'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
    'depth': trial.suggest_int('depth', 4, 6),
        'l2_leaf_reg': trial.suggest_float('l2_leaf_reg', 0.1, 10),
        'bagging_temperature': trial.suggest_float('bagging_temperature', 0.1, 1.0),
        'random_strength': trial.suggest_float('random_strength', 0.1, 10),
        'random_seed': RANDOM_SEED,
        'verbose': 0,
    }
    # TimeSeriesSplit with 3 splits
    tscv = TimeSeriesSplit(n_splits=2)
    rmse_scores = []
    for train_idx, val_idx in tscv.split(X_train_sorted):
        X_tr, X_va = X_train_sorted[train_idx], X_train_sorted[val_idx]
        y_tr, y_va = y_train_sorted[train_idx], y_train_sorted[val_idx]
        model = cb.CatBoostRegressor(**params)
        model.fit(X_tr, y_tr, eval_set=(X_va, y_va), early_stopping_rounds=30, verbose=False)
        pred = model.predict(X_va)
        rmse = np.sqrt(mean_squared_error(y_va, pred))
        rmse_scores.append(rmse)
    return np.mean(rmse_scores)

study = optuna.create_study(direction='minimize')
study.optimize(objective, n_trials=5, timeout=600)
best_params = study.best_params
logger.info("Best CatBoost hyperparameters:", best_params)
```

```python
# 4. Train final model with best params on full training set, early stopping on val ---
final_model = cb.CatBoostRegressor(
    **best_params,
    random_seed=RANDOM_SEED,
    verbose=100
)
final_model.fit(
    X_train_sorted, y_train_sorted,
    eval_set=(X_val_full, y_val),
    early_stopping_rounds=30,
)

# Evaluate on test
test_pred = final_model.predict(X_test_full)
test_rmse = np.sqrt(mean_squared_error(y_test, test_pred))
test_mae = mean_absolute_error(y_test, test_pred)
logger.info(f"Test RMSE: {test_rmse:.4f}, Test MAE: {test_mae:.4f}")

# Q‑error
def q_error(y_true: np.ndarray, y_pred: np.ndarray, percentiles: list = None) -> dict:
    if percentiles is None:
        percentiles = [50, 90, 95, 99]
    ratios = np.maximum(y_pred / y_true, y_true / y_pred)
    return {f'Q_error_p{p}': np.percentile(ratios, p) for p in percentiles}

q_errors = q_error(y_test, test_pred)
logger.info("Q‑errors:", q_errors)
```

```python
# 6. Ablation studies
# Helper to train and evaluate CatBoost with given feature subset indices
def train_eval_subset(feature_indices, label) -> float:
    X_tr = X_train_sorted[:, feature_indices] if USE_SORTED else X_train_full[:, feature_indices]
    X_va = X_val_full[:, feature_indices]
    X_te = X_test_full[:, feature_indices]
    model = cb.CatBoostRegressor(**best_params, random_seed=RANDOM_SEED, verbose=0)
    model.fit(X_tr, y_train_sorted if USE_SORTED else y_train,
              eval_set=(X_va, y_val), early_stopping_rounds=30)
    pred = model.predict(X_te)
    rmse = np.sqrt(mean_squared_error(y_test, pred))
    logger.info(f"Ablation [{label}]: Test RMSE = {rmse:.4f}")
    return rmse

# Full features index (all)
all_idx = list(range(NUM_TAB + NUM_TEXT))
base_rmse = test_rmse  # already computed

# Ablation: without text
no_text_idx = list(range(NUM_TAB))
rmse_no_text = train_eval_subset(no_text_idx, "without text")

# Ablation: without temporal features
# We keep all features except temporal ones (remove those indices from tabular, keep text)
temp_removed_tab = [i for i in range(NUM_TAB) if i not in temporal_feat_indices]
no_temp_idx = temp_removed_tab + list(range(NUM_TAB, NUM_TAB + NUM_TEXT))
rmse_no_temp = train_eval_subset(no_temp_idx, "without temporal")

ablation_results = {
    "full_rmse": base_rmse,
    "no_text_rmse": rmse_no_text,
    "no_temporal_rmse": rmse_no_temp
}
```

```python
# ============================================================
# 4.4 Hybrid Recommender – SVD + BERT Content
# ============================================================
RATING_THRESHOLD = 7.0          # above this -> positive feedback
COLD_START_THRESH = 5           # users with < this many interactions in train

# 1. Build interaction matrix (same as in 3.4)
inter_sql = f"""
    SELECT f.nconst AS user_id, f.tconst AS item_id,
           d.average_rating AS rating, d.start_year AS year
    FROM fact_performance f TABLESAMPLE SYSTEM ({SAMPLE_PERCENT} PERCENT) REPEATABLE({RANDOM_SEED})
    JOIN dim_title d ON f.tconst = d.tconst
    WHERE f.category IN ('actor','actress','director')
      AND d.average_rating IS NOT NULL AND d.start_year IS NOT NULL
"""
df_inter = con.execute(inter_sql).df()
logger.info(f"Interaction rows: {len(df_inter)}")

# Map IDs
user_ids = df_inter['user_id'].unique()
item_ids = df_inter['item_id'].unique()
user2idx = {u: i for i, u in enumerate(user_ids)}
item2idx = {it: i for i, it in enumerate(item_ids)}
df_inter['user_idx'] = df_inter['user_id'].map(user2idx)
df_inter['item_idx'] = df_inter['item_id'].map(item2idx)
num_users = len(user2idx)
num_items = len(item2idx)
logger.info(f"Users: {num_users}, Items: {num_items}")

# Temporal split
train_df = df_inter[df_inter['year'] <= TRAIN_YEAR_MAX]
val_df   = df_inter[(df_inter['year'] >= VAL_YEAR_MIN) & (df_inter['year'] <= VAL_YEAR_MAX)]
test_df  = df_inter[df_inter['year'] >= TEST_YEAR_MIN]
```

```python
# 2. Load title embeddings (DistilBERT) and align with item_idx

# Load base_features.parquet to get tconst order for embeddings
# (we know embeddings are in same row order as base_features)
base_path = PROCESSED_DIR / 'base_features.parquet'
df_base = pd.read_parquet(base_path, columns=['tconst'])  # only need tconst
tconst_order = df_base['tconst'].values

# Map tconst to embedding index
tconst_to_emb_idx = {tconst: idx for idx, tconst in enumerate(tconst_order)}

# Load embeddings (if sharded, we'd merge them; here we assume a single file)
embeddings = load_title_embeddings(PROCESSED_DIR)   # shape (n_titles, 768)

# Create a mapping from item_idx to embedding vector
item_emb_dict = {}   # item_idx -> embedding vector (or None if not found)
for item_id, idx in item2idx.items():
    if item_id in tconst_to_emb_idx:
        item_emb_dict[idx] = embeddings[tconst_to_emb_idx[item_id]]
    else:
        item_emb_dict[idx] = None
# Only keep items with embeddings
valid_items = [idx for idx, emb in item_emb_dict.items() if emb is not None]
logger.info(f"Items with embeddings: {len(valid_items)} / {num_items}")

# Filter interactions to only items that have embeddings
def filter_items(df, min_year, max_year) -> pd.DataFrame:
    return df[df['item_idx'].isin(valid_items)]

train_df = filter_items(train_df)
val_df = filter_items(val_df)
test_df = filter_items(test_df)

# Re‑index users that appear in the filtered set
active_users = set(train_df['user_idx']) | set(val_df['user_idx']) | set(test_df['user_idx'])
user_id_map = {u: i for i, u in enumerate(sorted(active_users))}
train_df['user_idx'] = train_df['user_idx'].map(user_id_map)
val_df['user_idx'] = val_df['user_idx'].map(user_id_map)
test_df['user_idx'] = test_df['user_idx'].map(user_id_map)
num_users_final = len(user_id_map)
logger.info(f"Final users: {num_users_final}, items: {len(valid_items)}")
```

```python
# 3. SVD (collaborative) model
reader = Reader(rating_scale=(1, 10))
train_svd = Dataset.load_from_df(train_df[['user_idx','item_idx','rating']], reader)
trainset = train_svd.build_full_trainset()

svd = SVD(n_factors=50, n_epochs=20, biased=True, random_state=RANDOM_SEED)
svd.fit(trainset)
logger.info("SVD trained.")
```

```python
# 4. Content‑based user embedding (average of high‑rated items in train)
user_embeddings = {}   # user_idx -> vector (768,)
for user, group in train_df[train_df['rating'] >= RATING_THRESHOLD].groupby('user_idx'):
    item_embeds = [item_emb_dict[i] for i in group['item_idx'] if i in item_emb_dict]
    if item_embeds:
        user_embeddings[user] = np.mean(item_embeds, axis=0)

# For users with no high‑rated items, use the global average item embedding
global_item_emb = np.mean([item_emb_dict[i] for i in valid_items if i in item_emb_dict], axis=0)
# Add global embedding to user dict for missing users
for u in train_df['user_idx'].unique():
    if u not in user_embeddings:
        user_embeddings[u] = global_item_emb
for u in val_df['user_idx'].unique():
    if u not in user_embeddings:
        user_embeddings[u] = global_item_emb
for u in test_df['user_idx'].unique():
    if u not in user_embeddings:
        user_embeddings[u] = global_item_emb

# 5. Compute scores for validation and test sets
def compute_scores(user_emb, item_emb_dict, svd_model) -> np.ndarray:
    svd_scores = []
    content_scores = []
    for _, row in df.iterrows():
        u, i = int(row['user_idx']), int(row['item_idx'])
        # SVD prediction (raw rating)
        try:
            s = svd.predict(u, i).est
        except Exception as e:
            logger.warning(f"SVD predict failed: {e}")
            s = 3.0   # default if unknown
        svd_scores.append(s)
        # Content cosine similarity
        u_emb = user_embeddings.get(u, global_item_emb)
        i_emb = item_emb_dict.get(i, global_item_emb)
        sim = cosine_similarity([u_emb], [i_emb])[0][0]
        content_scores.append(sim)
    return np.array(svd_scores), np.array(content_scores)

svd_val, cont_val = compute_scores(val_df)
svd_test, cont_test = compute_scores(test_df)
```

```python
# 6. Hybrid model (logistic regression on binary relevance)

# Binary target: rating >= RATING_THRESHOLD
y_val_bin = (val_df['rating'].values >= RATING_THRESHOLD).astype(int)
y_test_bin = (test_df['rating'].values >= RATING_THRESHOLD).astype(int)

# Normalize features for LR
scaler = StandardScaler()
X_val_comb = np.column_stack([svd_val, cont_val])
X_val_scaled = scaler.fit_transform(X_val)
X_test_comb = np.column_stack([svd_test, cont_test])
X_test_scaled = scaler.transform(X_test)

# Train logistic regression
lr = LogisticRegression(random_state=RANDOM_SEED)
lr.fit(X_val_scaled, y_val_bin)

# Hybrid scores (probability)
hybrid_val_proba = lr.predict_proba(X_val_scaled)[:, 1]
hybrid_test_proba = lr.predict_proba(X_test_scaled)[:, 1]
```

```python
# 7. Evaluation: cold‑start Precision@10, Recall@10
def precision_recall_at_k(df: pd.DataFrame, scores: np.ndarray, k: int = 10) -> tuple:
    """For each user, rank items by scores (descending), evaluate Precision@k and Recall@k."""
    df = df.copy()
    df['score'] = scores
    precisions = []
    recalls = []
    for user, group in df.groupby('user_idx'):
        # Get top-k items
        topk = group.nlargest(k, 'score')
        true_pos = topk['rating'] >= RATING_THRESHOLD
        # Total relevant items for this user in the test set
        rel_items = (group['rating'] >= RATING_THRESHOLD).sum()
        prec = true_pos.sum() / k if k > 0 else 0
        rec = true_pos.sum() / rel_items if rel_items > 0 else 0
        precisions.append(prec)
        recalls.append(rec)
    return np.mean(precisions), np.mean(recalls)

# Identify cold‑start users (interactions < threshold in training)
train_user_counts = train_df.groupby('user_idx').size()
cold_users = train_user_counts[train_user_counts < COLD_START_THRESH].index
cold_test_df = test_df[test_df['user_idx'].isin(cold_users)]
logger.info(f"Cold‑start test users: {len(cold_test_df['user_idx'].unique())}")

if len(cold_users) < 30:
    logger.warning(f"Only {len(cold_users)} cold-start users (< 30). Skipping cold-start gate.")

# Get positions of cold-start users within the full test set
cold_positions = test_df.index.get_indexer(cold_test_df.index)

# Hybrid evaluation
hyb_prec, hyb_rec = precision_recall_at_k(cold_test_df, hybrid_test_proba[cold_positions], k=10)
logger.info(f"Hybrid Cold‑Start Precision@10: {hyb_prec:.4f}, Recall@10: {hyb_rec:.4f}")

# Ablation: only SVD
svd_prec, svd_rec = precision_recall_at_k(cold_test_df, svd_test[cold_positions], k=10)
logger.info(f"SVD‑only Cold‑Start Precision@10: {svd_prec:.4f}, Recall@10: {svd_rec:.4f}")

# Ablation: only content (use content score directly)
cont_prec, cont_rec = precision_recall_at_k(cold_test_df, cont_test[cold_positions], k=10)
logger.info(f"Content‑only Cold‑Start Precision@10: {cont_prec:.4f}, Recall@10: {cont_rec:.4f}")
```
