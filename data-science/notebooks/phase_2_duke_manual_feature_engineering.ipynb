```python
# 0.3. Connect to DuckDB and register Gold Parquet views

# Either use ':memory' for temporary or 'imdb_gold.db'
con = duckdb.connect(database = 'imdb_gold.db')
con.execute("PRAGMA threads=4")
con.execute("PRAGMA memory_limit='8GB'")

# Use TABLESAMPLE if in development mode
sample_clause = f"TABLESAMPLE SYSTEM ({SAMPLE_PERCENT} PERCENT) REPEATABLE(42)" if DEVELOPMENT_MODE else ""
# Note: We will apply sampling at query time to keep views full (to preserve join flexibility)

views = {
    'dim_title':            DATA_DIR / 'dim_title.parquet',
    'dim_person':           DATA_DIR / 'dim_person.parquet',
    'fact_title_rating':    DATA_DIR / 'fact_title_rating.parquet',
    'fact_performance':     DATA_DIR / 'fact_performance.parquet',
    'fact_episode':         DATA_DIR / 'fact_episode.parquet',
    'fact_title_principal': DATA_DIR / 'fact_title_principal.parquet',
}

for name, path in views.items():
    if path.exists():
        con.execute(f"CREATE OR REPLACE VIEW {name} AS SELECT * FROM read_parquet('{path.as_posix()}')")
        logger.info(f"✓ {name}")
    else:
        logger.info(f"✗ Missing: {path}")

# Quick row counts
for name in views:
    cnt = con.execute(f"SELECT count(*) FROM {name}").fetchone()[0]
    logger.info(f"{name:25s} → {cnt:>12,} rows")

# Dev mode: materialize sampled temp tables once (avoids repeated TABLESAMPLE I/O)
if DEVELOPMENT_MODE:
    logger.info(f"\n⏳ Creating sampled temp tables ({SAMPLE_PERCENT}% sample) …")
    sampled = {}
    for name in ('dim_title', 'fact_performance', 'fact_episode', 'dim_person'):
        sm_name = f"{name}_sm"
        con.execute(f"""
            CREATE OR REPLACE TEMP TABLE {sm_name} AS
            SELECT * FROM {name} TABLESAMPLE SYSTEM ({SAMPLE_PERCENT} PERCENT) REPEATABLE(42)
        """)
        cnt = con.execute(f"SELECT count(*) FROM {sm_name}").fetchone()[0]
        sampled[name] = sm_name
        logger.info(f"  ✓ {name} → {sm_name} ({cnt:,} rows)")
    # Override views to read from sampled temp tables (materialized, fast)
    for orig, sm in sampled.items():
        con.execute(f"CREATE OR REPLACE VIEW {orig} AS SELECT * FROM {sm}")
    # Clear sample_clause so SQL does not re-sample (views are already sampled)
    sample_clause = ""
    logger.info("✅ Dev mode: views → sampled temp tables, sample_clause cleared")
```

### Phase 1: Feature Engineering & Temporal Split ###

```python
# ============================================================
# 1.1 Extract base features (title‑level)
# ============================================================

# SQL: assemble title features from dim_title, fact_performance, fact_episode
# We do NOT use dim_person directly; person counts come from fact_performance.
# fact_title_rating is not needed because dim_title already holds the latest rating.
base_feat_sql = f"""
WITH
-- 1. sampled titles (core grain)
sampled_titles AS (
    SELECT * FROM dim_title {sample_clause}
    WHERE title_type IN ('movie', 'tvSeries', 'tvMiniSeries', 'tvMovie', 'short')
      AND start_year IS NOT NULL
),

-- 2. aggregated person counts per title
perf_counts AS (
    SELECT
        tconst,
        COUNT(*)                        AS num_persons,
        COUNT(DISTINCT nconst)          AS unique_persons,
        COUNT(CASE WHEN category = 'actor'          THEN 1 END) AS actor_count,
        COUNT(CASE WHEN category = 'actress'        THEN 1 END) AS actress_count,
        COUNT(CASE WHEN category = 'director'       THEN 1 END) AS director_count,
        COUNT(CASE WHEN category = 'writer'         THEN 1 END) AS writer_count,
        COUNT(CASE WHEN category = 'producer'       THEN 1 END) AS producer_count,
        COUNT(CASE WHEN category = 'composer'       THEN 1 END) AS composer_count,
        COUNT(CASE WHEN category = 'editor'         THEN 1 END) AS editor_count,
        COUNT(CASE WHEN category = 'cinematographer' THEN 1 END) AS cinematographer_count,
        COUNT(CASE WHEN category = 'self'           THEN 1 END) AS self_count,
        COUNT(CASE WHEN category = 'archive_footage' THEN 1 END) AS archive_footage_count,
        COUNT(CASE WHEN category = 'archive_sound'  THEN 1 END) AS archive_sound_count
    FROM fact_performance {sample_clause}
    GROUP BY tconst
),

-- 3. episode‑level aggregations for series (join back to series_key)
episode_counts AS (
    SELECT
        series_key,
        COUNT(*)                    AS series_episode_count,
        AVG(avg_rating)            AS series_avg_episode_rating,
        MIN(season_number)         AS min_season,
        MAX(season_number)         AS max_season
    FROM (
        SELECT e.series_key, e.season_number, d.average_rating AS avg_rating
        FROM fact_episode e {sample_clause}
        JOIN dim_title d {sample_clause} ON e.episode_key = d.tconst
        WHERE d.average_rating IS NOT NULL
          AND d.start_year <= {VAL_YEAR_MAX}  -- prevent future data leakage
    )
    GROUP BY series_key
)

-- 4. final SELECT
SELECT
    t.tconst,
    t.title_type,
    t.primary_title,
    CAST(t.is_adult AS INTEGER)     AS is_adult,
    t.start_year,
    t.end_year,
    t.runtime_minutes,
    t.genre_list,
    t.average_rating,
    t.num_votes,
    t.region_list,
    t.language_list,
    t.aka_count,
    t.popularity_segment,
    t.rating_bucket,
    COALESCE(pc.num_persons, 0)    AS num_persons,
    COALESCE(pc.unique_persons, 0) AS unique_persons,
    COALESCE(pc.actor_count, 0)    AS actor_count,
    COALESCE(pc.actress_count, 0)  AS actress_count,
    COALESCE(pc.director_count, 0) AS director_count,
    COALESCE(pc.writer_count, 0)   AS writer_count,
    COALESCE(pc.producer_count, 0) AS producer_count,
    COALESCE(pc.composer_count, 0) AS composer_count,
    COALESCE(pc.editor_count, 0)   AS editor_count,
    COALESCE(pc.cinematographer_count, 0) AS cinematographer_count,
    COALESCE(pc.self_count, 0)     AS self_count,
    COALESCE(ec.series_episode_count, 0)  AS series_episode_count,
    ec.series_avg_episode_rating,
    CARDINALITY(string_split(COALESCE(t.genre_list, ''), ',')) AS genre_cnt,
    ec.min_season,
    ec.max_season
FROM sampled_titles t
LEFT JOIN perf_counts pc ON t.tconst = pc.tconst
LEFT JOIN episode_counts ec ON t.tconst = ec.series_key
"""

# Execution (dev: direct Dataframe; full: export to Parquet to avoid memory overflow)
if DEVELOPMENT_MODE:
    # Limit rows to keep memory low and allow quick inspection
    base_feat_sql += " LIMIT 250000"
    df_features = con.execute(base_feat_sql).df()
    df_features.to_parquet(PROCESSED_DIR / 'base_features.parquet', index=False)
    logger.info(f"✅ Development mode – extracted {df_features.shape[0]} rows × {df_features.shape[1]} columns")
else:
    # Full run: write directly to Parquet (DuckDB can stream this)
    out_path = (PROCESSED_DIR / 'base_features.parquet').as_posix()
    con.execute(f"COPY ({base_feat_sql}) TO '{out_path}' (FORMAT PARQUET)")
    # Optionally load a peek
    df_features = con.execute(f"SELECT * FROM read_parquet('{out_path}') LIMIT 5").df()
    total_rows = con.execute(f"SELECT count(*) FROM read_parquet('{out_path}')").fetchone()[0]
    logger.info(f"✅ Full mode – exported {total_rows} rows to base_features.parquet")

display(df_features.sample(n = 5))
```

```python
# ============================================================
# 1.2 Temporal aggregations
# ============================================================

# ------------------------------------------------------------
# Step 1: Genre‑year statistics (average rating & votes per genre per year)
# ------------------------------------------------------------
genre_year_sql = f"""
CREATE OR REPLACE TEMP TABLE genre_year_stats AS
SELECT
    genre,
    start_year,
    AVG(average_rating) AS avg_rating_genre_year,
    AVG(num_votes)      AS avg_votes_genre_year,
    COUNT(*)            AS genre_year_title_count
FROM (
    SELECT
        unnest(string_split(genre_list, ',')) AS genre,
        start_year,
        average_rating,
        num_votes
    FROM dim_title {sample_clause}
    WHERE genre_list IS NOT NULL
      AND start_year IS NOT NULL
      AND average_rating IS NOT NULL
)
GROUP BY genre, start_year
"""

con.execute(genre_year_sql)
logger.info("✅ genre_year_stats created")

# ------------------------------------------------------------
# Step 2: Person career metrics (directors & writers)
# ------------------------------------------------------------
# 2a. Person debut year
con.execute(f"""
CREATE OR REPLACE TEMP TABLE person_debut AS
SELECT
    nconst,
    category,
    MIN(start_year) AS debut_year
FROM fact_performance f {sample_clause}
JOIN dim_title d {sample_clause} ON f.tconst = d.tconst
WHERE d.start_year IS NOT NULL
  AND f.category IN ('director', 'writer')
GROUP BY nconst, category
""")
logger.info("✅ person_debut created")

# 2b. For each person and each year they were active, compute experience stats
con.execute(f"""
CREATE OR REPLACE TEMP TABLE person_year_stats AS
WITH person_credits AS (
    SELECT
        f.nconst,
        f.category,
        d.start_year
    FROM fact_performance f {sample_clause}
    JOIN dim_title d {sample_clause} ON f.tconst = d.tconst
    WHERE d.start_year IS NOT NULL
      AND f.category IN ('director', 'writer')
),
credit_counts AS (
    SELECT
        nconst,
        category,
        start_year,
        COUNT(*) OVER (PARTITION BY nconst, category ORDER BY start_year
                       ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS cumulative_credits,
        COUNT(*) OVER (PARTITION BY nconst, category ORDER BY start_year
                       RANGE BETWEEN 5 PRECEDING AND CURRENT ROW)      AS recent_5yr_credits
    FROM person_credits
)
SELECT DISTINCT
    pc.nconst,
    pc.category,
    pc.start_year,
    pd.debut_year,
    pc.start_year - pd.debut_year AS career_length,
    pc.cumulative_credits,
    pc.recent_5yr_credits
FROM credit_counts pc
JOIN person_debut pd ON pc.nconst = pd.nconst AND pc.category = pd.category
""")
logger.info("✅ person_year_stats created")

# ------------------------------------------------------------

# ------------------------------------------------------------
# Step 4: Aggregate per title – director/writer career features
# ------------------------------------------------------------
# For each title, get the directors and writers, join their year stats, average.
con.execute(f"""
CREATE OR REPLACE TEMP TABLE title_crew_features AS
WITH title_crew AS (
    SELECT
        f.tconst,
        f.nconst,
        f.category,
        d.start_year AS title_year
    FROM fact_performance f {sample_clause}
    JOIN dim_title d {sample_clause} ON f.tconst = d.tconst
    WHERE f.category IN ('director', 'writer')
      AND d.start_year IS NOT NULL
),
crew_with_stats AS (
    SELECT
        tc.tconst,
        tc.category,
        AVG(ps.career_length)           AS avg_career_length,
        MAX(ps.career_length)           AS max_career_length,
        AVG(ps.cumulative_credits)      AS avg_experience,
        AVG(ps.recent_5yr_credits)      AS avg_recent_activity
    FROM title_crew tc
    JOIN person_year_stats ps
      ON tc.nconst = ps.nconst
     AND tc.category = ps.category
     AND tc.title_year = ps.start_year   -- stats at the title's release year
    GROUP BY tc.tconst, tc.category
)
SELECT
    tconst,
    MAX(CASE WHEN category = 'director' THEN avg_career_length END)   AS dir_avg_career_len,
    MAX(CASE WHEN category = 'director' THEN max_career_length END)   AS dir_max_career_len,
    MAX(CASE WHEN category = 'director' THEN avg_experience END)      AS dir_avg_experience,
    MAX(CASE WHEN category = 'director' THEN avg_recent_activity END) AS dir_avg_recent_activity,
    MAX(CASE WHEN category = 'writer'   THEN avg_career_length END)   AS wri_avg_career_len,
    MAX(CASE WHEN category = 'writer'   THEN max_career_length END)   AS wri_max_career_len,
    MAX(CASE WHEN category = 'writer'   THEN avg_experience END)      AS wri_avg_experience,
    MAX(CASE WHEN category = 'writer'   THEN avg_recent_activity END) AS wri_avg_recent_activity
FROM crew_with_stats
GROUP BY tconst
""")
logger.info("✅ title_crew_features created")

# ------------------------------------------------------------
# Step 5: Assemble the complete temporal feature table
# ------------------------------------------------------------
temporal_feat_sql = """
SELECT
    b.tconst,
    -- director career features
    cf.dir_avg_career_len,
    cf.dir_max_career_len,
    cf.dir_avg_experience,
    cf.dir_avg_recent_activity,
    -- writer career features
    cf.wri_avg_career_len,
    cf.wri_max_career_len,
    cf.wri_avg_experience,
    cf.wri_avg_recent_activity
FROM base_features b
LEFT JOIN title_crew_features   cf ON b.tconst = cf.tconst
"""

if DEVELOPMENT_MODE:
    # We already have df_features from 1.1, but to keep consistency we'll re‑query using the base_features CTE
    # First we need a base_features table (from 1.1 output). We'll use the df_features tconst to filter.
    # Easier: we already have df_features; we can directly join via DuckDB by registering df_features as a temporary table.
    con.execute("CREATE OR REPLACE TEMP TABLE base_features AS SELECT * FROM df_features")
    df_temporal = con.execute(temporal_feat_sql).df()
    logger.info(f"✅ Dev mode – temporal features shape: {df_temporal.shape}")
else:
    # Full mode: export to Parquet using the base_features.parquet file as source
    # We'll assume base_features.parquet exists; if not, we could use dim_title directly.
    base_parquet = (PROCESSED_DIR / 'base_features.parquet').as_posix()
    out_parquet  = (PROCESSED_DIR / 'temporal_features.parquet').as_posix()
    con.execute(f"""
        COPY (
            WITH base_features AS (
                SELECT * FROM read_parquet('{base_parquet}')
            )
            {temporal_feat_sql}
        ) TO '{out_parquet}' (FORMAT PARQUET)
    """)
    total_rows = con.execute(f"SELECT count(*) FROM read_parquet('{out_parquet}')").fetchone()[0]
    logger.info(f"✅ Full mode – exported {total_rows} rows to temporal_features.parquet")

if DEVELOPMENT_MODE:
    display(df_temporal.sample(n = 5))
```

```python
# ============================================================
# 1.3 Text embeddings (DistilBERT)
# ============================================================

# Config
EMBEDDING_DIM = 768          # DistilBERT-base
BATCH_SIZE = 64              # adjust based on GPU memory
SHARD_SIZE = 500_000         # rows per .npy shard in full mode
MODEL_NAME = 'distilbert-base-uncased'

# Load tokenizer & model once
tokenizer = DistilBertTokenizer.from_pretrained(MODEL_NAME)
model = DistilBertModel.from_pretrained(MODEL_NAME)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
model.eval()
logger.info(f"Model loaded on {device}")
```

```python
def embed_text_batch(texts, tokenizer, model, device, batch_size=BATCH_SIZE)(texts: list, batch_size: int = 32) -> np.ndarray:
    """Return numpy array of [CLS] embeddings for a list of texts."""
    all_embs = []
    for i in range(0, len(texts), batch_size):
        batch = texts[i:i+batch_size]
        inputs = tokenizer(batch, return_tensors='pt', padding=True,
                           truncation=True, max_length=32).to(device)
        with torch.no_grad():
            outputs = model(**inputs)
        # CLS token embedding
        cls_emb = outputs.last_hidden_state[:, 0, :].cpu().numpy()
        all_embs.append(cls_emb)
    return np.vstack(all_embs)

if DEVELOPMENT_MODE:
    # Dev mode: Embed the sampled df_features
    logger.info("Embedding titles (development sample) …")
    titles = df_features['primary_title'].fillna('').tolist()
    start_t = time.time()
    embeddings = embed_text_batch(titles, tokenizer, model, device, BATCH_SIZE)
    logger.info(f"Done in {time.time()-start_t:.1f}s, shape {embeddings.shape}")
    np.save(PROCESSED_DIR / 'title_embeddings.npy', embeddings)
    logger.info("Saved title_embeddings.npy")

else:
    # Full mode: Stream from base_features.parquet and save in shards
    base_parquet = (PROCESSED_DIR / 'base_features.parquet').as_posix()
    total_rows = con.execute(f"SELECT count(*) FROM read_parquet('{base_parquet}')").fetchone()[0]
    num_shards = math.ceil(total_rows / SHARD_SIZE)
    logger.info(f"Full mode: {total_rows} rows, will write {num_shards} shard(s) of {SHARD_SIZE} each")

    for shard_idx in range(num_shards):
        offset = shard_idx * SHARD_SIZE
        # Fetch this shard's titles as a list
        chunk_df = con.execute(f"""
            SELECT primary_title
            FROM read_parquet('{base_parquet}')
            ORDER BY tconst
            LIMIT {SHARD_SIZE} OFFSET {offset}
        """).df()
        titles = chunk_df['primary_title'].fillna('').tolist()
        logger.info(f"Shard {shard_idx+1}/{num_shards}: {len(titles)} titles …")
        start_t = time.time()
        embs = embed_text_batch(titles, tokenizer, model, device, BATCH_SIZE)
        logger.info(f"  → embedded in {time.time()-start_t:.1f}s")
        shard_path = PROCESSED_DIR / f'title_embeddings_shard_{shard_idx:04d}.npy'
        np.save(shard_path, embs)
        logger.info(f"  → saved {shard_path}")

    # Optionally also save a metadata file with shard info
    shard_info = {
        'total_rows': total_rows,
        'shard_size': SHARD_SIZE,
        'num_shards': num_shards,
        'embedding_dim': EMBEDDING_DIM,
        'model': MODEL_NAME
    }
    with open(PROCESSED_DIR / 'embedding_shards_meta.json', 'w') as f:
        json.dump(shard_info, f)
    logger.info("All shards saved. Metadata written to embedding_shards_meta.json")
```

```python
# ============================================================
# 1.4 Categorical encoding & numeric scaling
# ============================================================

# 1. Assemble merged feature DataFrame
if DEVELOPMENT_MODE:
    # dev: merge df_features (from 1.1) with df_temporal (from 1.2)
    df_merged = df_features.merge(df_temporal, on='tconst', how='left')
    logger.info(f"Dev merge: {df_merged.shape[0]} rows")
else:
    # full: load parquet files and join
    base_parquet = (PROCESSED_DIR / 'base_features.parquet').as_posix()
    temporal_parquet = (PROCESSED_DIR / 'temporal_features.parquet').as_posix()
    df_base = con.execute(f"SELECT * FROM read_parquet('{base_parquet}')").df()
    df_temp = con.execute(f"SELECT * FROM read_parquet('{temporal_parquet}')").df()
    df_merged = df_base.merge(df_temp, on='tconst', how='left')
    logger.info(f"Full merge: {df_merged.shape[0]} rows") 
```

```python
# Save the merged features for further training
output_path = PROCESSED_DIR / "merged_features.parquet"
df_merged.to_parquet(output_path, index=False)

logger.info(f"Saved to {output_path}")
```

```python
# 2. Define feature columns
numeric_cols = [
    'start_year', 'runtime_minutes', 'average_rating', 'num_votes',
    'num_persons', 'unique_persons', 'actor_count', 'director_count',
    'writer_count', 'composer_count', 'genre_cnt',
    'dir_avg_career_len', 'dir_max_career_len', 'dir_avg_experience', 'dir_avg_recent_activity',
    'wri_avg_career_len', 'wri_max_career_len', 'wri_avg_experience', 'wri_avg_recent_activity'
]
# Simple categoricals (to be one‑hot encoded)
categorical_cols = ['title_type', 'is_adult']

# Multi‑label columns (will be binarized separately)
multi_label_cols = ['genre_list', 'region_list']

# Ensure columns exist in df_merged
numeric_cols = [c for c in numeric_cols if c in df_merged.columns]
categorical_cols = [c for c in categorical_cols if c in df_merged.columns]
multi_label_cols = [c for c in multi_label_cols if c in df_merged.columns]

# Fill missing numeric with median (using training split later) and categorical with 'unknown'
# We'll do the imputation inside the ColumnTransformer pipeline.

# 3. Temporal split
train_mask = df_merged['start_year'] <= TRAIN_YEAR_MAX
val_mask   = (df_merged['start_year'] >= VAL_YEAR_MIN) & (df_merged['start_year'] <= VAL_YEAR_MAX)
test_mask  = df_merged['start_year'] >= TEST_YEAR_MIN

logger.info(f"Train: {train_mask.sum()}, Val: {val_mask.sum()}, Test: {test_mask.sum()}")
```

```python
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='constant', fill_value='missing')),
    ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_cols),
        ('cat', categorical_transformer, categorical_cols)
    ]
)

# Fit on training data
X_train_raw = df_merged.loc[train_mask, numeric_cols + categorical_cols]
preprocessor.fit(X_train_raw)

# Transform all sets
X_train_proc = preprocessor.transform(df_merged.loc[train_mask, numeric_cols + categorical_cols])
X_val_proc   = preprocessor.transform(df_merged.loc[val_mask, numeric_cols + categorical_cols])
X_test_proc  = preprocessor.transform(df_merged.loc[test_mask, numeric_cols + categorical_cols])

# Extract scaler for standalone artifact

def safe_minmax(X):
    """Min-max normalization with epsilon to prevent ZeroDivisionError."""
    mn = X.min(axis=0)
    mx = X.max(axis=0)
    return (X - mn) / (mx - mn + 1e-8)

X_train_proc = safe_minmax(X_train_proc)
X_val_proc   = safe_minmax(X_val_proc)
X_test_proc  = safe_minmax(X_test_proc)

scaler = preprocessor.named_transformers_['num'].named_steps['scaler']

logger.info(f"Transformed feature shapes: train {X_train_proc.shape}, val {X_val_proc.shape}, test {X_test_proc.shape}")
```

```python
# 5. Multi‑label binarizers

# We'll create binarizers for genre_list and region_list.
# First, convert comma‑delimited strings to lists.
def to_list(series)(s: str) -> list:
    return series.fillna('').apply(lambda x: [item.strip() for item in x.split(',') if item.strip()] if isinstance(x, str) else [])

mlbs = {}
y_train_dict = {}
y_val_dict = {}
y_test_dict = {}

for col in multi_label_cols:
    if col not in df_merged.columns:
        continue
    # Extract lists
    lists_train = to_list(df_merged.loc[train_mask, col])
    lists_val   = to_list(df_merged.loc[val_mask, col])
    lists_test  = to_list(df_merged.loc[test_mask, col])
    
    mlb = MultiLabelBinarizer()
    y_train_dict[col] = mlb.fit_transform(lists_train)
    y_val_dict[col]   = mlb.transform(lists_val)
    y_test_dict[col]  = mlb.transform(lists_test)
    mlbs[col] = mlb
    logger.info(f"MultiLabelBinarizer for '{col}': {len(mlb.classes_)} classes")
```

```python
# 6. Save preprocessor and binarizers
joblib.dump(preprocessor, PROCESSED_DIR / 'preprocessor.joblib')
joblib.dump(scaler, PROCESSED_DIR / 'scaler.joblib')
for col, mlb in mlbs.items():
    joblib.dump(mlb, PROCESSED_DIR / f'{col}_mlb.joblib')
logger.info("✅ Preprocessor and MultiLabelBinarizers saved.")
```

```python
# Optionally, also save the raw train/val/test DataFrames with indices for later use
split_info = df_merged[['tconst', 'start_year']].copy()
split_info['split'] = 'none'
split_info.loc[train_mask, 'split'] = 'train'
split_info.loc[val_mask, 'split'] = 'val'
split_info.loc[test_mask, 'split'] = 'test'
split_info.to_parquet(PROCESSED_DIR / 'temporal_split.parquet', index=False)
logger.info("✅ Temporal split saved.")
```

```python
# ============================================================
# 1.5 Temporal split creation
# ============================================================

# Ensure we have a DataFrame with tconst & start_year
if 'df_merged' in locals() and 'tconst' in df_merged.columns and 'start_year' in df_merged.columns:
    split_source = df_merged[['tconst', 'start_year']].copy()
    logger.info("Using in‑memory df_merged for split creation.")
else:
    # Fallback: load from base_features.parquet (full mode or if df_merged missing)
    base_path = PROCESSED_DIR / 'base_features.parquet'
    if base_path.exists():
        split_source = con.execute(
            f"SELECT tconst, start_year FROM read_parquet('{base_path.as_posix()}')"
        ).df()
        logger.info("Loaded from base_features.parquet for split creation.")
    else:
        raise FileNotFoundError("Neither df_merged nor base_features.parquet available. Run 1.1 first.")

# Create split label
split_source['split'] = 'none'
split_source.loc[split_source['start_year'] <= TRAIN_YEAR_MAX, 'split'] = 'train'
split_source.loc[(split_source['start_year'] >= VAL_YEAR_MIN) & 
                 (split_source['start_year'] <= VAL_YEAR_MAX), 'split'] = 'val'
split_source.loc[split_source['start_year'] >= TEST_YEAR_MIN, 'split'] = 'test'

# Keep only necessary columns
split_indices = split_source[['tconst', 'split']]

# Save
out_path = PROCESSED_DIR / 'split_indices.parquet'
split_indices.to_parquet(out_path, index=False)
logger.info(f"✅ Saved {out_path} ({len(split_indices)} rows)")

# Quick summary
logger.info(split_indices['split'].value_counts())
```

```python
# ============================================================
# 1.6 Assemble final feature matrices (genre & rating pillars)
# ============================================================

# Helper: Align and concatenate text embeddings
def get_text_embeddings_for_split(mask, embeddings_array, df_merged)(split_name: str, df: pd.DataFrame) -> np.ndarray:
    """Return text embeddings for rows in mask, assuming embeddings_array is aligned with df_merged."""
    return embeddings_array[mask.values, :]

# 1. Load or use in‑memory data (dev mode)
if DEVELOPMENT_MODE:
    # We should have df_merged, embeddings, X_train_proc, etc. from previous sections.
    # If any is missing, raise an error.
    required = ['df_merged', 'embeddings', 'X_train_proc', 'X_val_proc', 'X_test_proc']
    for var in required:
        if var not in globals():
            raise RuntimeError(f"{var} missing. Please run sections 1.1–1.4 first.")
    logger.info("Using in‑memory development data.")
else:
    # Full mode: load all data from parquet/npy/joblib
    logger.info("Full mode: loading data from disk …")
    # base and temporal features
    base_path = PROCESSED_DIR / 'base_features.parquet'
    temp_path = PROCESSED_DIR / 'temporal_features.parquet'
    df_base = con.execute(f"SELECT * FROM read_parquet('{base_path.as_posix()}')").df()
    df_temp = con.execute(f"SELECT * FROM read_parquet('{temp_path.as_posix()}')").df()
    df_merged = df_base.merge(df_temp, on='tconst', how='left')
    
    # Load preprocessor and apply
if not (PROCESSED_DIR / 'preprocessor.joblib').exists(): raise FileNotFoundError(f"Required artifact missing: {PROCESSED_DIR / 'preprocessor.joblib'}. Run the upstream notebook first.")
    preprocessor = joblib.load(PROCESSED_DIR / 'preprocessor.joblib')
    # columns are those used by the preprocessor
    numeric_cols = [c for c in preprocessor.transformers_[0][2] if c in df_merged.columns]
    categorical_cols = [c for c in preprocessor.transformers_[1][2] if c in df_merged.columns]
    X_all = preprocessor.transform(df_merged[numeric_cols + categorical_cols])
    
    # Load split indices
    split_df = con.execute(f"SELECT * FROM read_parquet('{(PROCESSED_DIR / 'temporal_split.parquet').as_posix()}')").df()
    # Ensure alignment: split_df tconst order may differ; we need to align with df_merged order.
    # We'll merge to get the split column in df_merged order.
    df_merged_with_split = df_merged.merge(split_df, on='tconst', how='left')
    train_mask = df_merged_with_split['split'] == 'train'
    val_mask   = df_merged_with_split['split'] == 'val'
    test_mask  = df_merged_with_split['split'] == 'test'
    
    # Load text embeddings (possibly sharded)
    meta_path = PROCESSED_DIR / 'embedding_shards_meta.json'
    if meta_path.exists():
        with open(meta_path, 'r') as f:
            shard_meta = json.load(f)
        # For simplicity, we assume we can memory-map all shards into one array.
        # Caution: may exceed RAM if total embeddings are huge. Consider using memmap or loading on-the-fly.
        all_embeddings = []
        for shard_idx in range(shard_meta['num_shards']):
            shard_path = PROCESSED_DIR / f'title_embeddings_shard_{shard_idx:04d}.npy'
            shard = np.load(shard_path)
            all_embeddings.append(shard)
        embeddings = np.vstack(all_embeddings)
    else:
        # Single embeddings file
if not (PROCESSED_DIR / 'title_embeddings.npy').exists(): raise FileNotFoundError(f"Required artifact missing: {PROCESSED_DIR / 'title_embeddings.npy'}. Run the upstream notebook first.")
        embeddings = np.load(PROCESSED_DIR / 'title_embeddings.npy')
    
    # Align processed tabular features to the same row order as df_merged
    # X_all is in the same row order as df_merged (we transformed df_merged directly).
    # Assign
    X_train_proc = X_all[train_mask]
    X_val_proc   = X_all[val_mask]
    X_test_proc  = X_all[test_mask]
```

```python
# 2. Merge tabular & text features for each split
X_train_text = get_text_embeddings_for_split(train_mask, embeddings, df_merged)
X_val_text   = get_text_embeddings_for_split(val_mask, embeddings, df_merged)
X_test_text  = get_text_embeddings_for_split(test_mask, embeddings, df_merged)

X_train = np.hstack([X_train_proc, X_train_text])
X_val   = np.hstack([X_val_proc, X_val_text])
X_test  = np.hstack([X_test_proc, X_test_text])

logger.info(f"Final feature shapes: train {X_train.shape}, val {X_val.shape}, test {X_test.shape}")
```

```python
# 3. Targets for genre classification (multi-label)

# to_list and mlbs['genre_list'] are in scope from Cell 12
mlb_genre = mlbs['genre_list']
y_train_genre = mlb_genre.transform(to_list(df_merged.loc[train_mask, 'genre_list']))
y_val_genre   = mlb_genre.transform(to_list(df_merged.loc[val_mask, 'genre_list']))
y_test_genre  = mlb_genre.transform(to_list(df_merged.loc[test_mask, 'genre_list']))

logger.info(f"Genre target shapes: train {y_train_genre.shape}, val {y_val_genre.shape}, test {y_test_genre.shape}")
```

```python
# 4. Targets for rating regression

# Average_rating column exists; we need to filter rows where rating is not NaN (and possibly num_votes > some threshold)
rating_mask_train = df_merged.loc[train_mask, 'average_rating'].notna()
rating_mask_val   = df_merged.loc[val_mask, 'average_rating'].notna()
rating_mask_test  = df_merged.loc[test_mask, 'average_rating'].notna()

y_train_rating = df_merged.loc[train_mask, 'average_rating'][rating_mask_train].values.astype(float)
y_val_rating   = df_merged.loc[val_mask, 'average_rating'][rating_mask_val].values.astype(float)
y_test_rating  = df_merged.loc[test_mask, 'average_rating'][rating_mask_test].values.astype(float)

# Corresponding feature subsets for rating (only rows with rating)
X_train_rating = X_train[rating_mask_train.values]
X_val_rating   = X_val[rating_mask_val.values]
X_test_rating  = X_test[rating_mask_test.values]

logger.info(f"Rating target shapes: train {y_train_rating.shape}, val {y_val_rating.shape}, test {y_test_rating.shape}")
```

```python
# 5. Save all matrices
# Genre classification
np.save(PROCESSED_DIR / 'X_train_genre.npy', X_train)
np.save(PROCESSED_DIR / 'X_val_genre.npy', X_val)
np.save(PROCESSED_DIR / 'X_test_genre.npy', X_test)
np.save(PROCESSED_DIR / 'y_train_genre.npy', y_train_genre)
np.save(PROCESSED_DIR / 'y_val_genre.npy', y_val_genre)
np.save(PROCESSED_DIR / 'y_test_genre.npy', y_test_genre)

# Rating regression
np.save(PROCESSED_DIR / 'X_train_rating.npy', X_train_rating)
np.save(PROCESSED_DIR / 'X_val_rating.npy', X_val_rating)
np.save(PROCESSED_DIR / 'X_test_rating.npy', X_test_rating)
np.save(PROCESSED_DIR / 'y_train_rating.npy', y_train_rating)
np.save(PROCESSED_DIR / 'y_val_rating.npy', y_val_rating)
np.save(PROCESSED_DIR / 'y_test_rating.npy', y_test_rating)

# Feature column names for later use
# Tabular feature names from the preprocessor's one‑hot encoder
preprocessor = joblib.load(PROCESSED_DIR / 'preprocessor.joblib')
num_feats = preprocessor.transformers_[0][2]  # numeric column names
cat_encoder = preprocessor.named_transformers_['cat'].named_steps['onehot']
cat_feats = list(cat_encoder.get_feature_names_out(categorical_cols))
all_tab_feats = list(num_feats) + cat_feats
# text embedding feature names
text_feats = [f'text_emb_{i}' for i in range(EMBEDDING_DIM)]
all_features = all_tab_feats + text_feats

feature_info = {
    'tabular_features': all_tab_feats,
    'text_features': text_feats,
    'total_features': len(all_features)
}
with open(PROCESSED_DIR / 'feature_columns.json', 'w') as f:
    json.dump(feature_info, f)

logger.info("✅ All final feature matrices and targets saved.")
# Close DuckDB connection
con.close()
```
